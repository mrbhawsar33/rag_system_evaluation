```python
import os
from dotenv import load_dotenv

load_dotenv()
HF_TOKEN = os.getenv("HF_TOKEN")
```

```python
"""
Run Mistral-7B-Instruct on Hugging Face *hosted* inference (NOT locally),
using the chat-completion / conversational API.

Install:
  pip install -U huggingface_hub

Set token (recommended):
  Windows PowerShell:
    setx HF_TOKEN "hf_xxx..."
  Then restart your terminal.
"""

import os
from huggingface_hub import InferenceClient

HF_TOKEN = os.getenv("HF_TOKEN")
if not HF_TOKEN:
    raise RuntimeError("Missing HF_TOKEN env var. Set it to your Hugging Face access token.")

MODEL_ID = "mistralai/Mistral-7B-Instruct-v0.2"

# Create a hosted inference client
# NOTE: If you omit provider, HF may auto-pick a provider (like featherless-ai).
# Using chat_completion avoids the 'text-generation not supported' error for such mappings.
client = InferenceClient(api_key=HF_TOKEN)

messages = [
    {"role": "system", "content": "You are a helpful assistant. Answer concisely."},
    {"role": "user", "content": "what is 2+2? Reply with just the number."},
]

resp = client.chat_completion(
    model=MODEL_ID,
    messages=messages,
    max_tokens=16,
    temperature=0.0,
)

print(resp.choices[0].message["content"].strip())
```

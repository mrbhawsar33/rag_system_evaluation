##   Step 3: Simple Question — Ask Function

```python
import ollama

MODEL = 'qwen3:4b'

def ask(question, system=None):
    """Ask simple one-shot question """
    messages = []
    if system:
        messages.append({'role': 'system', 'content': system})
    messages.append({'role': 'user', 'content': question})
    
    response = ollama.chat(model=MODEL, messages=messages)
    return response['message']['content']

# Test karo!
print(ask('What is Python? Explain in 2 sentences.'))
```

##  Step 4: Streaming Response (Real-time output)

```python
import ollama

def ask_stream(question):
    """Real-time streaming — word by word output"""
    print(' Response:')
    stream = ollama.chat(
        model='qwen3:4b',
        messages=[{'role': 'user', 'content': question}],
        stream=True
    )
    for chunk in stream:
        print(chunk['message']['content'], end='', flush=True)
    print()

ask_stream('Explain what is a neural network in simple terms.')
```

##  Step 5: Multi-turn Chat (Conversation History)

```python
import ollama

class Chat:
    def __init__(self, system=None):
        self.history = []
        if system:
            self.history.append({'role': 'system', 'content': system})

    def send(self, message):
        self.history.append({'role': 'user', 'content': message})
        resp = ollama.chat(model='qwen3:4b', messages=self.history)
        reply = resp['message']['content']
        self.history.append({'role': 'assistant', 'content': reply})
        print(f' You: {message}')
        print(f' Bot: {reply}')
        print('-' * 50)
        return reply

    def reset(self):
        self.history = []
        print(' Chat reset!')


# Chat session =
bot = Chat(system='You are a helpful assistant. Be concise.')

bot.send('What is machine learning?')
bot.send('Give me one real world example of it.')
```
